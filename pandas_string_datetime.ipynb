```python
import pandas as pd

data = {
    'Name':  ['  arun  ', 'PRIYA', 'karthik', 'DIVYA  ', 'raj'],
    'Email': ['arun@gmail.com', 'PRIYA@YAHOO.COM',
              'karthik@gmail.com', 'divya@hotmail.com', 'raj@gmail.com'],
    'City':  ['chennai', 'MUMBAI', 'Delhi', 'CHENNAI', 'mumbai']
}
df = pd.DataFrame(data)
print(df)
```

```python
# Fix spaces
df['Name'] = df['Name'].str.strip()      # remove spaces
print(df['Name'])

# Fix case
df['Name'] = df['Name'].str.title()      # Proper Case
print(df['Name'])
print("+++++++++_____========================================_______++++++")
df['City'] = df['City'].str.upper()      # UPPER CASE
print(df['City'])
print("+++++++++_____========================================_______++++++")
df['Email'] = df['Email'].str.lower()    # lower case
print(df['Email'])
print("+++++++++_____========================================_______++++++")

print(df)
print("++===========================================================++")

# Check contains
print(df['Email'].str.contains('gmail')) # True/False

# Extract domain from email
df['Domain'] = df['Email'].str.split('@').str[1]
print(df['Domain'])
print("++===========================================================++")


# Replace values
df['City'] = df['City'].str.replace('MUMBAI', 'Mumbai')
print(df)
print("++===========================================================++")

# Length of string
df['Name_Length'] = df['Name'].str.len()
print(df)
print("++===========================================================++")
```

```python
data = {
    'Name':       ['Arun', 'Priya', 'Karthik', 'Divya', 'Raj'],
    'JoinDate':   ['2023-01-15', '2022-06-20', '2023-03-10',
                   '2021-11-05', '2023-07-25'],
    'Salary':     [15000, 25000, 12000, 22000, 18000]
}
df = pd.DataFrame(data)

# Convert string to datetime
df['JoinDate'] = pd.to_datetime(df['JoinDate'])
print(df.JoinDate)
# Extract parts
df['Year']  = df['JoinDate'].dt.year
print(df.Year)
df['Month'] = df['JoinDate'].dt.month
print(df.Month)

df['Day']   = df['JoinDate'].dt.day
print(df.Day)
# Day of week
df['DayOfWeek'] = df['JoinDate'].dt.day_name()
print(df.DayOfWeek)
print(df)
```

```python
data = {
    'Name':      ['  arun kumar  ', 'PRIYA S', 'karthik',
                  'DIVYA  ', 'raj'],
    'Email':     ['Arun@Gmail.Com', 'PRIYA@YAHOO.COM',
                  'karthik@gmail.com', 'divya@hotmail.com',
                  'raj@gmail.com'],
    'City':      ['chennai', 'MUMBAI', 'Delhi',
                  'CHENNAI', 'mumbai'],
    'JoinDate':  ['2023-01-15', '2022-06-20', '2023-03-10',
                  '2021-11-05', '2023-07-25'],
    'Salary':    [15000, 25000, 12000, 22000, 18000]
}
df = pd.DataFrame(data)

# Q1: Remove spaces from Name column
df['Name']=df['Name'].str.strip()
# Q2: Convert all Names to title case (Arun Kumar)
df['Name']=df['Name'].str.title()
# Q3: Convert all emails to lowercase
df['Email']=df['Email'].str.lower()
# Q4: Extract domain from email (gmail.com, yahoo.com)
df['Domain']=df['Email'].str.split('@').str[1]
# Q5: Convert City to uppercase
df['City']=df['City'].str.upper()
# Q6: Convert JoinDate to datetime
df['JoinDate']=pd.to_datetime(df['JoinDate'])

# Q7: Extract Year and Month from JoinDate
df["year"] = df["JoinDate"].dt.year
df["month"] = df["JoinDate"].dt.month
# Q8: Find which day of week each person joined
df['DayOfWeek']=df['JoinDate'].dt.day_name()
# Q9: Add column 'Gmail_User' → True if email has gmail
df['Gmail_User'] = df['Email'].str.contains('gmail')


# Q10: Find average salary of Gmail users vs others

#      Hint: groupby Gmail_User column!\
print(df.groupby('Gmail_User')['Salary'].mean())
```
